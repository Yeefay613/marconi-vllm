```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from IPython.display import display

sns.set_theme(style="whitegrid", context="talk")

paths = {
    "20GB KV / 20GB SSM": "prefix_cache_20kv_20ssm_sharegpt.csv",
    "20GB KV / 80GB SSM": "prefix_cache_20kv_80ssm_sharegpt.csv",
}

dfs = []
for budget, path in paths.items():
    tmp = pd.read_csv(path)
    tmp["budget"] = budget
    dfs.append(tmp)

df = pd.concat(dfs, ignore_index=True)

df["token_hit_pct"] = 100 * df["token_hit_rate"]
df["request_hit_pct"] = 100 * df["request_hit_rate"]
df["total_used_gb"] = df["kv_used_gb"] + df["ssm_used_gb"]
df["total_budget_gb"] = df["kv_capacity_gb"] + df["ssm_capacity_gb"]

display(df.head())
print(df.shape)
```

```python
# Best configs by cache type and budget.

best = (
    df.sort_values("token_hit_rate", ascending=False)
      .groupby(["budget", "cache_type"], as_index=False)
      .first()
)

cols = [
    "budget",
    "cache_type",
    "kv_block_size",
    "ssm_checkpoint_interval",
    "token_hit_pct",
    "request_hit_pct",
    "avg_hit_tokens_per_hit",
    "kv_used_gb",
    "ssm_used_gb",
    "total_used_gb",
    "evictions",
]

display(
    best[cols].style.format({
        "token_hit_pct": "{:.2f}%",
        "request_hit_pct": "{:.2f}%",
        "avg_hit_tokens_per_hit": "{:.1f}",
        "kv_used_gb": "{:.2f}",
        "ssm_used_gb": "{:.2f}",
        "total_used_gb": "{:.2f}",
        "evictions": "{:,.0f}",
    })
)
```

```python
best.to_csv('large_capacity_best_config.csv')
```

```python
# Plot 1: Hybrid hit rates across checkpoint interval and KV block size.

hybrid = df[df["cache_type"] == "hybrid_intersection"].copy()

fig, axes = plt.subplots(1, 2, figsize=(15, 5), sharey=False)

sns.lineplot(
    data=hybrid,
    x="ssm_checkpoint_interval",
    y="token_hit_pct",
    hue="budget",
    style="kv_block_size",
    marker="o",
    ax=axes[0],
)

axes[0].set_title("Hybrid Token Hit Rate")
axes[0].set_xlabel("SSM checkpoint interval")
axes[0].set_ylabel("Token hit rate (%)")

sns.lineplot(
    data=hybrid,
    x="ssm_checkpoint_interval",
    y="request_hit_pct",
    hue="budget",
    style="kv_block_size",
    marker="o",
    ax=axes[1],
)

axes[1].set_title("Hybrid Request Hit Rate")
axes[1].set_xlabel("SSM checkpoint interval")
axes[1].set_ylabel("Request hit rate (%)")

plt.suptitle("Hybrid Cache Hit Rates by Budget and Granularity", y=1.04)
plt.tight_layout()
plt.show()
```

```python
# Plot 2: Best achievable hit rate by cache type.
# This compares KV-only, SSM-only, and hybrid using each one's best config.

best_for_plot = (
    df.sort_values("token_hit_rate", ascending=False)
      .groupby(["budget", "cache_type"], as_index=False)
      .first()
)

plt.figure(figsize=(10, 5))
sns.barplot(
    data=best_for_plot,
    x="budget",
    y="token_hit_pct",
    hue="cache_type",
)

plt.title("Best Token Hit Rate by Cache Type")
plt.xlabel("")
plt.ylabel("Best token hit rate (%)")
plt.legend(title="Cache type")
plt.tight_layout()
plt.show()

display(
    best_for_plot[[
        "budget",
        "cache_type",
        "kv_block_size",
        "ssm_checkpoint_interval",
        "token_hit_pct",
        "request_hit_pct",
        "kv_used_gb",
        "ssm_used_gb",
        "evictions",
    ]].style.format({
        "token_hit_pct": "{:.2f}%",
        "request_hit_pct": "{:.2f}%",
        "kv_used_gb": "{:.2f}",
        "ssm_used_gb": "{:.2f}",
        "evictions": "{:,.0f}",
    })
)
```

```python
# Plot 3: Memory consumption vs token hit rate.
# See whether extra SSM memory is buying meaningful hit-rate improvement.

plot_df = df[df["cache_type"].isin(["hybrid_intersection"])].copy()

plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=plot_df,
    x="ssm_used_gb",
    y="token_hit_pct",
    hue="budget",
    style="cache_type",
    size="ssm_checkpoint_interval",
    sizes=(80, 220),
    alpha=0.85,
)

plt.title("SSM Memory Consumption vs Token Hit Rate")
plt.xlabel("SSM memory used (GB)")
plt.ylabel("Token hit rate (%)")
plt.legend(bbox_to_anchor=(1.04, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=plot_df,
    x="ssm_used_gb",
    y="token_hit_pct",
    hue="budget",
    style="kv_block_size",
    size="ssm_checkpoint_interval",
    sizes=(80, 220),
    alpha=0.85,
)

plt.title("SSM Memory Consumption vs Token Hit Rate")
plt.xlabel("SSM memory used (GB)")
plt.ylabel("Token hit rate (%)")
plt.legend(bbox_to_anchor=(1.04, 1), loc="upper left")
plt.tight_layout()
plt.show()
```
